```python
### Add function definition
def add(a,b):
    result = a + b
    return result
```

```python
### Test
add(2, 6)
```

```python
### Sum list using add

def sum_list(values):
    total = 0
    for value in values:
        total = add(total, value)
    return total

sum_list([1, 2, 3])
```

```python
sum_list([10,5,8])
```
